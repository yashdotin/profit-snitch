```python
outlier_summary = []

for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]

    outlier_summary.append({
        'Column': col,
        'Outlier Count': len(outliers),
        'Outlier %': round(len(outliers) / len(df) * 100, 2),
        'Lower Bound': round(lower_bound, 2),
        'Upper Bound': round(upper_bound, 2)
    })

outlier_df = pd.DataFrame(outlier_summary)
outlier_df
```

```python

df['Rating'] = df['Rating'].fillna(df['Rating'].median())


df['Gender'] = df['Gender'].fillna(df['Gender'].mode()[0])


df['Category'] = df['Category'].fillna(df['Category'].mode()[0])


```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'], format='%d-%m-%y')
df['Order_Month'] = df['Order_Date'].dt.month
df['Order_Weekday'] = df['Order_Date'].dt.dayofweek

df = df.drop(columns=['Order_ID', 'Order_Date'])
```

```python
from sklearn.model_selection import train_test_split

y = df['Profit_Category']


X_with_profit = df.drop(columns=['Profit_Category'])

X_train_wp, X_test_wp, y_train_wp, y_test_wp = train_test_split(
    X_with_profit, y, test_size=0.2, random_state=42, stratify=y
)


```

```python
X_without_profit = df.drop(columns=['Profit', 'Profit_Category'])

X_train_np, X_test_np, y_train_np, y_test_np = train_test_split(
    X_without_profit, y, test_size=0.2, random_state=42, stratify=y
)
```

```python
print("With Profit:", X_train_wp.shape, X_test_wp.shape)
print("Without Profit:", X_train_np.shape, X_test_np.shape)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC

models = {
    'LogisticRegression': LogisticRegression(class_weight='balanced', random_state=42),
    'DecisionTreeClassifier': DecisionTreeClassifier(class_weight='balanced', random_state=42),
    'RandomForestClassifier': RandomForestClassifier(n_estimators=200, class_weight='balanced', random_state=42),
    'KNeighborsClassifier': KNeighborsClassifier(n_neighbors=5),
    'SVM': SVC(probability=True, class_weight='balanced', random_state=42),
    'GradientBoostingClassifier': GradientBoostingClassifier(random_state=42)
}
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import classification_report, roc_auc_score,confusion_matrix

def train_and_evaluate(X_train, X_test, y_train, y_test, label):
    scaler = StandardScaler()
    X_train_s = scaler.fit_transform(X_train)
    X_test_s = scaler.transform(X_test)

    results = {}
    for name, model in models.items():
        model.fit(X_train_s, y_train)
        pred = model.predict(X_test_s)
        auc = roc_auc_score(y_test, model.predict_proba(X_test_s)[:, 1])

        print(f"--- {label} | {name} ---")
        print(classification_report(y_test, pred))
        print(f"ROC-AUC: {auc:.4f}\n")

        results[name] = auc
    return results
```

```python
results_wp = train_and_evaluate(X_train_wp, X_test_wp, y_train_wp, y_test_wp, "WITH Profit")
```

```python
results_np = train_and_evaluate(X_train_np, X_test_np, y_train_np, y_test_np, "WITHOUT Profit")
```

```python
comparison = pd.DataFrame({
    'With Profit (leaky)': results_wp,
    'Without Profit (clean)': results_np
})
comparison['Gap'] = comparison['With Profit (leaky)'] - comparison['Without Profit (clean)']
comparison = comparison.sort_values('Without Profit (clean)', ascending=False)
comparison
```

```python
from sklearn.model_selection import cross_val_score

for name in ['LogisticRegression', 'RandomForestClassifier', 'GradientBoostingClassifier']:
    scores = cross_val_score(models[name], X_without_profit, y, cv=5, scoring='roc_auc')
    print(name, "-> Mean:", round(scores.mean(),4), "| Std:", round(scores.std(),4))
```

```python
scaler = StandardScaler()
X_train_s = scaler.fit_transform(X_train_np)
X_test_s = scaler.transform(X_test_np)

rf_final = RandomForestClassifier(n_estimators=200, class_weight='balanced', random_state=42)
rf_final.fit(X_train_s, y_train_np)
pred_final = rf_final.predict(X_test_s)

cm = confusion_matrix(y_test_np, pred_final)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
```

```python
importances = pd.Series(rf_final.feature_importances_, index=X_without_profit.columns)
importances = importances.sort_values(ascending=False)

plt.figure(figsize=(8,6))
sns.barplot(x=importances.values, y=importances.index)
plt.show()
```
